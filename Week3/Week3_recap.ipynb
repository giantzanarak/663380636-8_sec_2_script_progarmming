## 2. `for` Loop และ `range()` (1.5 นาที)

```python
# วนซ้ำ 5 ครั้ง (0 ถึง 4)
for i in range(5):
    print(f"Iteration {i}")

# range(start, stop) -- เริ่มที่ 1 ไปจนถึงก่อน 11
for num in range(1, 11):
    print(num)

# range(start, stop, step) -- นับทีละ 2
for even_num in range(0, 11, 2):
    print(even_num)

# วนซ้ำทีละตัวอักษรในสตริง
my_string = "Python"
for char in my_string:
    print(f"Character: {char}")

# ตัวอย่างการรวมค่า
total_sum = 0
for i in range(1, 6):
    total_sum += i          # เท่ากับ total_sum = total_sum + i
print(f"Sum of first 5 numbers: {total_sum}")   # 15
```

**จำไว้:** `range(stop)` เริ่มที่ 0 เสมอ และ **ไม่รวม** ค่า `stop`

## 3. `while` Loop (1 นาที)

ทำงานซ้ำไปเรื่อย ๆ **ตราบใดที่เงื่อนไขยังเป็น True** — ต้องมี 3 ส่วนเสมอ: กำหนดค่าเริ่มต้น, เงื่อนไข, และการอัปเดตค่า (ไม่งั้นจะเป็น infinite loop!)

```python
count = 5
while count > 0:
    print(f"Count: {count}")
    count -= 1              # ขาดบรรทัดนี้ = loop ไม่รู้จบ
print("Countdown finished!")

# ใช้ตรวจสอบ input ของผู้ใช้
user_input = ""
while not (user_input == "yes" or user_input == "no"):
    user_input = input("Please enter 'yes' or 'no': ").lower()
```

## 4. `break` และ `continue` (1 นาที)

- **`break`** — หยุด loop ทันที
- **`continue`** — ข้ามรอบปัจจุบัน แล้วไปรอบถัดไป

```python
secret_number = 7
attempts = 0
max_attempts = 3

while attempts < max_attempts:
    guess = int(input(f"Guess the number (1-10) - Attempt {attempts + 1}/{max_attempts}: "))
    attempts += 1
    if guess == secret_number:
        print("Congratulations! You guessed it!")
        break                # เจอคำตอบแล้ว ออกจาก loop ทันที
    elif guess < secret_number:
        print("Too low!")
    else:
        print("Too high!")
else:                        # else ของ while: ทำงานเมื่อ loop จบโดยไม่ถูก break
    print(f"Sorry, out of attempts. The secret number was {secret_number}.")

# continue: ข้ามเลขคี่ พิมพ์เฉพาะเลขคู่
for i in range(1, 11):
    if i % 2 != 0:
        continue
    print(f"Even number: {i}")
```

## 5. Nested Loops — Loop ซ้อน Loop (1 นาที)

Loop ชั้นในจะวนจนครบทุกรอบ **ก่อน** loop ชั้นนอกจะขยับไปรอบถัดไป — ใช้บ่อยกับตาราง กริด และลวดลาย

```python
# สามเหลี่ยมมุมฉาก
height = 5
for row in range(1, height + 1):   # แถวที่ 1 ถึง 5
    for col in range(row):          # พิมพ์ดอกจันตามจำนวนแถว
        print("* ", end="")
    print()
# *
# * *
# * * *
# * * * *
# * * * * *
```

---"""
learning_log_week3_th.py
--------------------------
เครื่องมือบันทึกการเรียนรู้สัปดาห์ที่ 3: ต่อยอดจากสัปดาห์ที่แล้ว
คราวนี้ใช้ for loop เพื่อบันทึกได้หลายหัวข้อในเซสชันเดียว
และใช้ while loop เพื่อตรวจสอบ input ให้ถูกต้องก่อนบันทึก

รันด้วยคำสั่ง:
    python learning_log_week3_th.py
"""

import logging
from datetime import datetime

logging.basicConfig(
    filename="learning_log.txt",
    level=logging.INFO,
    format="%(asctime)s | %(message)s",
    datefmt="%Y-%m-%d %H:%M",
)


def ask_confidence():
    """
    while loop ตรวจสอบ input: บังคับให้ผู้ใช้กรอกเลข 1-5 เท่านั้น
    ไม่งั้นจะถามซ้ำไปเรื่อย ๆ (เหมือนตัวอย่าง input validation ของสัปดาห์นี้)
    """
    confidence = -1
    while confidence < 1 or confidence > 5:
        raw = input("ระดับความมั่นใจ (1-5): ")
        if not raw.isdigit():
            print("กรุณากรอกเป็นตัวเลข 1-5 เท่านั้น")
            continue                     # ข้ามรอบนี้ ถามใหม่
        confidence = int(raw)
        if confidence < 1 or confidence > 5:
            print("ต้องอยู่ระหว่าง 1 ถึง 5 เท่านั้น ลองใหม่อีกครั้ง")
    return confidence


def log_multiple_topics():
    """
    for loop: ให้ผู้ใช้บันทึกได้หลายหัวข้อในเซสชันเดียว
    พิมพ์ 'done' เพื่อออกก่อนครบจำนวนก็ได้ (break)
    """
    max_topics = 5   # กันไม่ให้วนไม่รู้จบถ้าผู้ใช้ลืมพิมพ์ done
    logged_count = 0

    for i in range(max_topics):
        topic = input(f"\nหัวข้อที่ {i + 1} (หรือพิมพ์ 'done' เพื่อจบ): ")
        if topic.lower() == "done":
            break                        # ออกจาก for loop ทันที

        confidence = ask_confidence()
        entry = f"Topic: {topic} | Confidence: {confidence}/5"
        logging.info(entry)
        print(f"บันทึกแล้ว: {entry}")
        logged_count += 1
    else:
        # else ของ for: ทำงานเมื่อวนครบ max_topics โดยไม่ถูก break
        print(f"\nครบจำนวนสูงสุด {max_topics} หัวข้อต่อเซสชันแล้ว")

    return logged_count


def main():
    print("=== บันทึกการเรียนรู้ สัปดาห์ที่ 3: Loops ===")
    print(f"วันที่วันนี้: {datetime.now():%Y-%m-%d}")

    total = log_multiple_topics()
    print(f"\nสรุป: บันทึกไปทั้งหมด {total} หัวข้อในเซสชันนี้")


if __name__ == "__main__":
    main()
