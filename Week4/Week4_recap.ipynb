## 2. List — ลำดับข้อมูลที่แก้ไขได้ (4 นาที)

List คือลำดับข้อมูลที่ **เรียงลำดับ (ordered)**, **แก้ไขได้ (mutable)**, และ **มีค่าซ้ำกันได้** เขียนด้วยวงเล็บเหลี่ยม `[]`

### การสร้างและเข้าถึงข้อมูล

```python
empty_list = []
fruits = ["apple", "banana", "cherry", "date"]

# indexing: เริ่มนับที่ 0 / index ติดลบนับจากท้าย
print(fruits[0])       # apple
print(fruits[-1])      # date (ตัวสุดท้าย)

# slicing: list[start:end:step]  -- end ไม่ถูกนับรวม
print(fruits[1:3])     # ['banana', 'cherry']
print(fruits[1:])      # ['banana', 'cherry', 'date']
print(fruits[-2:])     # ['cherry', 'date']
print(fruits[::2])     # ['apple', 'cherry']  -- ทุกตัวที่ 2
```

### การแก้ไข List (Mutability)

```python
fruits[1] = "blueberry"          # เปลี่ยนค่าตรง index

fruits.append("elderberry")      # เพิ่มท้ายลิสต์
fruits.insert(1, "grape")        # แทรกที่ตำแหน่งที่กำหนด
fruits.extend(["kiwi", "lemon"]) # รวมสมาชิกจากลิสต์อื่นเข้ามา

fruits.remove("cherry")          # ลบตามค่า (ตัวแรกที่เจอ)
popped = fruits.pop(2)           # ลบตาม index และคืนค่าที่ลบออกมา
del fruits[0]                    # ลบตาม index โดยตรง
# fruits.clear()                 # ล้างทั้งลิสต์ให้ว่างเปล่า
```

### เมธอด/ฟังก์ชันที่ใช้บ่อย

```python
numbers = [3, 1, 4, 1, 5, 9, 2]

print(len(numbers))          # 7 -- จำนวนสมาชิก
print(numbers.count(1))      # 2 -- นับจำนวนที่ค่าเท่ากับ 1
print(numbers.index(4))      # 2 -- ตำแหน่งของค่า 4 ตัวแรก

numbers.sort()                # เรียงลำดับ "แก้ไขตัวเดิม" (in-place)
numbers.reverse()             # กลับลำดับ "แก้ไขตัวเดิม" (in-place)

unsorted = [5, 2, 8, 1, 9]
new_sorted = sorted(unsorted) # ได้ "ลิสต์ใหม่" ไม่แตะของเดิม
```

⚠️ **ข้อควรระวัง:** `sort()`/`reverse()` แก้ไขลิสต์เดิมโดยตรงและคืนค่า `None` ในขณะที่ `sorted()` คืนลิสต์ใหม่โดยไม่แตะของเดิม — สับสนสองตัวนี้บ่อยมาก

### การวนซ้ำผ่าน List (ต่อยอดจากสัปดาห์ที่ 3)

```python
for fruit in fruits:
    print(f"I like {fruit}")

for i in range(len(fruits)):
    print(f"Fruit at index {i}: {fruits[i]}")
```

## 3. Tuple — ลำดับข้อมูลที่แก้ไขไม่ได้ (4 นาที)

Tuple คือลำดับข้อมูลที่ **เรียงลำดับเหมือน list** แต่ **แก้ไขไม่ได้ (immutable)** เขียนด้วยวงเล็บกลม `()`

```python
empty_tuple = ()
colors = ("red", "green", "blue", "yellow")

# ทูเพิลที่มีสมาชิกตัวเดียว ต้องมีจุลภาคเสมอ!
single_item_tuple = ("single",)      # นี่คือ tuple จริง
not_a_tuple = ("single")             # นี่แค่สตริงธรรมดาในวงเล็บ
print(type(single_item_tuple), type(not_a_tuple))
# <class 'tuple'> <class 'str'>

# indexing และ slicing ใช้แบบเดียวกับ list ทุกประการ
print(colors[0])        # red
print(colors[1:3])      # ('green', 'blue')
```

### สาธิต Immutability

```python
try:
    colors[0] = "orange"          # พยายามแก้ไขค่า
except TypeError as e:
    print(f"Error trying to modify tuple: {e}")

try:
    colors.append("purple")       # พยายามเพิ่มค่า -- tuple ไม่มีเมธอดนี้ด้วยซ้ำ
except AttributeError as e:
    print(f"Error trying to append to tuple: {e}")

try:
    del colors[0]                 # พยายามลบค่า
except TypeError as e:
    print(f"Error trying to delete from tuple: {e}")

print(colors)   # ไม่มีอะไรเปลี่ยนแปลงเลย
```

### Tuple Unpacking และการแปลงชนิดข้อมูล

```python
point = (10, 20)
x_coord, y_coord = point            # แตกค่า tuple ใส่ตัวแปรทีเดียว
print(x_coord, y_coord)             # 10 20

x, y, z = 1, 2, 3                   # ไม่ใส่วงเล็บก็ยังเป็น tuple packing

my_list = [1, 2, 3, 4]
as_tuple = tuple(my_list)           # list -> tuple

my_tuple = ("alpha", "beta", "gamma")
as_list = list(my_tuple)            # tuple -> list
```

**เมื่อไหร่ควรใช้ tuple แทน list?**
- ข้อมูลที่ไม่ควรถูกเปลี่ยนแปลง เช่น พิกัด GPS คงที่, ค่าสี RGB
- ใช้เป็น key ของ dictionary ได้ (เพราะต้อง immutable) — list ทำไม่ได้
- ประหยัดหน่วยความจำและเร็วกว่า list เล็กน้อย เหมาะกับข้อมูลชุดใหญ่ที่ตายตัว

## 4. ตารางเปรียบเทียบ List vs Tuple (1 นาที)

| คุณสมบัติ | List `[]` | Tuple `()` |
|---|---|---|
| แก้ไขได้ (mutable) | ✅ | ❌ |
| เรียงลำดับ | ✅ | ✅ |
| ซ้ำกันได้ | ✅ | ✅ |
| ใช้เป็น dictionary key ได้ | ❌ | ✅ |
| ความเร็ว/หน่วยความจำ | ช้ากว่าเล็กน้อย | เร็วและประหยัดกว่า |
| ตัวอย่างการใช้งาน | ตะกร้าสินค้าที่แก้ไขได้ | พิกัด GPS คงที่ |

### Lab 4.1: ระบบจัดการรายชื่อนักเรียน (ฝึก List)

```python
student_names = []

# เพิ่มนักเรียน
for _ in range(3):
    name = input("Enter student name: ")
    student_names.append(name)
print(student_names)

# ค้นหานักเรียน
search_name = input("Search for a student: ")
if search_name in student_names:
    print(f"Found at index {student_names.index(search_name)}")
else:
    print("Student not found.")

# ลบนักเรียน (ตรวจสอบก่อนเสมอ)
remove_name = input("Remove which student? ")
if remove_name in student_names:
    student_names.remove(remove_name)

student_names.sort()
print(f"Total students: {len(student_names)}")
```

### Lab 4.2: คำนวณระยะห่างระหว่างจุด (ฝึก Tuple)

```python
import math

point1 = (3, 4)
point2 = (6, 8)

x1, y1 = point1
x2, y2 = point2

distance = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)
print(f"Distance: {distance}")   # 5.0
```

## 6. Activity 4.1: List หรือ Tuple ดี? (2 นาที)

ลองตัดสินใจก่อนเปิดเฉลย — แต่ละสถานการณ์ควรใช้ List หรือ Tuple?

| สถานการณ์ | คำตอบ | เหตุผล |
|---|---|---|
| อุณหภูมิที่บันทึกทุกวันตลอดสัปดาห์ | List | ค่าจะถูกเพิ่มเรื่อย ๆ ทุกวัน |
| ค่าสี RGB ของสีหนึ่งสี เช่น `(255, 0, 0)` | Tuple | ค่าคงที่ ไม่ควรเปลี่ยน |
| รายการสินค้าที่เพิ่ม/ลบได้ในตะกร้า | List | ต้องเพิ่ม-ลบตลอดเวลา |
| ชื่อวันในสัปดาห์ตามลำดับ | Tuple | ลำดับตายตัว ไม่มีวันเปลี่ยน |
| คะแนนนักเรียนที่อาจถูกแก้ไขภายหลัง | List | ค่าต้องอัปเดตได้ |
| พิกัด GPS ของสถานที่ตายตัว | Tuple | ตำแหน่งไม่ควรเปลี่ยนแปลง |

"""
learning_log_week4_20min_th.py
---------------------------------
เวอร์ชันขยายของสมุดบันทึกการเรียนรู้สัปดาห์ที่ 4
ครอบคลุมแนวคิดเต็มรูปแบบของสัปดาห์นี้:
  - list (mutable) เก็บบันทึกระหว่างเซสชัน พร้อมเมนูวนซ้ำ (ต่อยอด Week 3)
  - append / remove / sort / index / count / len ของ list
  - แปลงเป็น tuple ตอนปิดเซสชัน เพื่อสาธิต immutability
  - tuple unpacking ตอนสรุปผล

รันด้วยคำสั่ง:
    python learning_log_week4_20min_th.py
"""

import logging
from datetime import datetime

logging.basicConfig(
    filename="learning_log.txt",
    level=logging.INFO,
    format="%(asctime)s | %(message)s",
    datefmt="%Y-%m-%d %H:%M",
)


def show_menu():
    print("\n--- เมนูบันทึกการเรียนรู้ ---")
    print("1) เพิ่มหัวข้อ")
    print("2) ลบหัวข้อ")
    print("3) ดูรายการทั้งหมด")
    print("4) เรียงลำดับตามความมั่นใจ")
    print("5) จบเซสชันและบันทึก")


def add_topic(entries):
    """entries เป็น list -- ใช้ append() เพราะแก้ไขได้ (mutable)"""
    topic = input("หัวข้อที่เรียน: ")
    confidence = int(input("ความมั่นใจ (1-5): "))
    entries.append((topic, confidence))     # เก็บเป็น tuple เล็ก ๆ ในลิสต์ใหญ่
    print(f"เพิ่มแล้ว: {topic}")


def remove_topic(entries):
    """ลบตามชื่อหัวข้อ -- ตรวจสอบก่อนด้วย 'in' เหมือน Lab 4.1"""
    target = input("ลบหัวข้อไหน: ")
    match = next((e for e in entries if e[0] == target), None)
    if match in entries:
        entries.remove(match)               # remove() ใช้ได้เพราะเป็น list
        print(f"ลบ '{target}' แล้ว")
    else:
        print("ไม่พบหัวข้อนี้ในรายการ")


def view_topics(entries):
    print(f"\nจำนวนหัวข้อทั้งหมด: {len(entries)}")     # len()
    for i, (topic, confidence) in enumerate(entries):
        print(f"  [{i}] {topic} -- {confidence}/5")


def sort_by_confidence(entries):
    entries.sort(key=lambda e: e[1], reverse=True)   # sort() แก้ไขลิสต์เดิม
    print("เรียงลำดับตามความมั่นใจ (มาก -> น้อย) แล้ว")
    view_topics(entries)


def finalize_session(entries_list):
    """
    ปิดผนึกด้วย tuple -- แสดงว่าข้อมูลของเซสชันนี้ล็อกแล้ว แก้ไขไม่ได้อีก
    """
    session_record = tuple(entries_list)      # list -> tuple

    for topic, confidence in session_record:  # tuple unpacking ในลูป
        entry_text = f"Topic: {topic} | Confidence: {confidence}/5"
        logging.info(entry_text)

    print(f"\nปิดเซสชันแล้ว บันทึกทั้งหมด {len(session_record)} หัวข้อลง learning_log.txt")
    return session_record


def main():
    print("=== บันทึกการเรียนรู้ สัปดาห์ที่ 4: Lists & Tuples (ฉบับ 20 นาที) ===")
    print(f"วันที่วันนี้: {datetime.now():%Y-%m-%d}")

    entries = []   # list ว่าง -- แก้ไขได้ตลอดจนกว่าจะจบเซสชัน

    while True:                        # เมนูวนซ้ำ ต่อยอดจาก while loop สัปดาห์ที่ 3
        show_menu()
        choice = input("เลือกเมนู: ")

        if choice == "1":
            add_topic(entries)
        elif choice == "2":
            remove_topic(entries)
        elif choice == "3":
            view_topics(entries)
        elif choice == "4":
            sort_by_confidence(entries)
        elif choice == "5":
            finalize_session(entries)
            break                       # ออกจาก while loop
        else:
            print("กรุณาเลือก 1-5 เท่านั้น")


if __name__ == "__main__":
    main()
