```python
# ============================================================
# Lab 6.1 - Simple Calculator Functions
# ------------------------------------------------------------
# โจทย์: สร้างฟังก์ชันคำนวณพื้นฐาน แล้วเรียกใช้ผ่านเมนูแบบ while loop
#        ฝึกเรื่อง parameter, return value, docstring และ default parameter
# ============================================================

# --- ข้อที่ 1: นิยามฟังก์ชันคำนวณ 4 ฟังก์ชัน พร้อม docstring อธิบายหน้าที่ ---

def add(a, b):
    """บวกเลขสองจำนวน แล้วคืนค่าผลบวกกลับไป"""
    return a + b


def subtract(a, b):
    """ลบเลขสองจำนวน (a - b) แล้วคืนค่าผลลบกลับไป"""
    return a - b


def multiply(a, b):
    """คูณเลขสองจำนวน แล้วคืนค่าผลคูณกลับไป"""
    return a * b


def divide(a, b):
    """หารเลขสองจำนวน (a / b) ถ้าตัวหารเป็น 0 จะคืนข้อความแจ้ง error แทนการหารจริง"""
    if b == 0:
        return "Error: Division by zero"   # กันโปรแกรมพังจาก ZeroDivisionError
    return a / b


# --- ข้อที่ 2 (Challenge): ฟังก์ชันยกกำลัง ที่ใช้ default parameter ---

def power(base, exponent=2):
    """ยกกำลัง base ด้วย exponent ถ้าไม่ส่ง exponent มา จะยกกำลังสองให้อัตโนมัติ"""
    return base ** exponent


# --- ข้อที่ 3: เมนูเลือกการคำนวณ ด้วย while loop (นำความรู้จาก Week 3 มาใช้) ---

while True:
    print("\n=== Simple Calculator ===")
    print("1. Add (+)")
    print("2. Subtract (-)")
    print("3. Multiply (*)")
    print("4. Divide (/)")
    print("5. Power (**)  <- Challenge: ใช้ default parameter")
    print("6. Exit")

    choice = input("Enter choice (1-6): ")

    # --- เมนูที่ 1-4: รับเลข 2 จำนวน แล้วเรียกฟังก์ชันตามที่ผู้ใช้เลือก ---
    if choice in ("1", "2", "3", "4"):
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))

        if choice == "1":
            result = add(num1, num2)
            symbol = "+"
        elif choice == "2":
            result = subtract(num1, num2)
            symbol = "-"
        elif choice == "3":
            result = multiply(num1, num2)
            symbol = "*"
        else:
            result = divide(num1, num2)
            symbol = "/"

        print(f"Result: {num1} {symbol} {num2} = {result}")

    # --- เมนูที่ 5: ยกกำลัง ถ้าไม่กรอกเลขชี้กำลัง จะใช้ค่า default (exponent=2) ---
    elif choice == "5":
        base = float(input("Enter base number: "))
        exponent_input = input("Enter exponent (กด Enter เลย ถ้าจะใช้ค่า default = 2): ")

        if exponent_input == "":
            result = power(base)                      # ไม่ส่ง exponent -> ใช้ค่า default = 2
            print(f"Result: {base} ** 2 = {result}   (ใช้ค่า default parameter)")
        else:
            exponent = float(exponent_input)
            result = power(base, exponent)            # ส่ง exponent เองตามที่ผู้ใช้กรอก
            print(f"Result: {base} ** {exponent} = {result}")

    # --- เมนูที่ 6: ออกจากโปรแกรม ---
    elif choice == "6":
        print("Exiting Calculator. Goodbye!")
        break

    else:
        print("Invalid choice! Please enter a number from 1 to 6.")
```

## Lab 6.2 - Custom Module and Standard Library Usage

**Part 1:** สร้างโมดูลของตัวเองชื่อ `my_utils.py` ที่มีฟังก์ชัน `greet(name)` และ `is_prime(number)`

cell ถัดไปใช้คำสั่งพิเศษ `%%writefile my_utils.py` (magic command) เพื่อ **เขียนไฟล์ `my_utils.py` ออกมาจริง ๆ**
ในโฟลเดอร์เดียวกับ notebook วิธีนี้ใช้ได้ทั้งใน VS Code และ Google Colab



```python
%%writefile my_utils.py
# ============================================================
# my_utils.py - Custom Module (Lab 6.2 Part 1)
# ------------------------------------------------------------
# โมดูลที่เขียนขึ้นเอง สำหรับเก็บฟังก์ชันที่ต้องเรียกใช้ซ้ำจากไฟล์อื่น
# วิธีใช้: import my_utils  แล้วเรียก my_utils.greet("ชื่อ")
# ============================================================


def greet(name):
    """แสดงข้อความทักทายผู้ใช้ตามชื่อที่ส่งเข้ามา"""
    print(f"Hello, {name}! Welcome to my utility functions.")


def is_prime(number):
    """ตรวจสอบจำนวนเฉพาะ คืนค่า True ถ้า number เป็นจำนวนเฉพาะ และ False ถ้าไม่ใช่"""
    if number < 2:              # 0, 1 และเลขติดลบ ไม่ใช่จำนวนเฉพาะ
        return False

    # หารตั้งแต่ 2 ถึงรากที่สองของ number พอ ถ้าหารลงตัวแปลว่าไม่ใช่จำนวนเฉพาะ
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False

    return True


PI_VALUE = 3.14159  # ตัวแปรค่าคงที่ที่ประกาศไว้ในโมดูล (โมดูลเก็บได้ทั้งฟังก์ชันและตัวแปร)
```
