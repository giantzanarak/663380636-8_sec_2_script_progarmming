# กิจกรรมที่ 6.1: Refactoring a Previous Lab Assignment

**โจทย์:** เลือกแลปเก่าจาก Week 2 (Conditionals) หรือ Week 3 (Loops) มา 1 ข้อ
แล้วหาส่วนของโค้ดที่ทำงานเป็นหน่วยเดียวกัน (logical unit) หรือเขียนซ้ำ ๆ
จากนั้นแยกออกมาเป็นฟังก์ชัน เพื่อให้อ่านง่ายและนำกลับมาใช้ซ้ำได้

**แลปที่เลือก:** Lab 2.1 - Number Classifier (Week 2)

## โค้ดเดิมก่อน Refactor (จาก Week2/lab2.ipynb)

```python
num = int(input("Enter a number: "))

if num > 0:
    sign = "บวก"
elif num < 0:
    sign = "ลบ"
else:
    sign = "ศูนย์"

if num % 2 == 0:
    parity = "คู่"
else:
    parity = "คี่"

print(f"ตรวจสอบเลข {num} เป็น {sign}")
print(f"เลข {num} เป็นเลข {parity}")
```

**ปัญหาของโค้ดเดิม**

1. ตรรกะการแยกประเภทเลขปนอยู่กับส่วนรับ input และส่วนแสดงผล อ่านแล้วแยกไม่ออกว่าส่วนไหนทำอะไร
2. ถ้าอยากตรวจเลขตัวที่ 2 ต้อง **copy โค้ดทั้งก้อนไปวางซ้ำ** ผิดหลัก DRY (Don't Repeat Yourself)
3. ทดสอบยาก เพราะตรรกะผูกกับ `input()` จะทดสอบทีต้องนั่งพิมพ์ทีละครั้ง

**แนวทางแก้:** แยกออกเป็น 2 ฟังก์ชันตามหน้าที่ คือ `is_positive_negative_zero(num)` กับ `is_even_odd(num)`
ให้แต่ละฟังก์ชัน **return ค่ากลับมา** (ไม่ print ข้างใน) จะได้นำผลลัพธ์ไปใช้ต่อได้อิสระ

````python
# ============================================================
# Activity 6.1 - Refactored: Number Classifier (จาก Lab 2.1)
# ------------------------------------------------------------
# แยกตรรกะออกเป็นฟังก์ชัน แต่ละฟังก์ชันรับ parameter และ return ค่ากลับ
# ============================================================

# --- ข้อที่ 1: ฟังก์ชันตรวจว่าเลขเป็นบวก ลบ หรือศูนย์ ---
def is_positive_negative_zero(num):
    """ตรวจสอบเครื่องหมายของตัวเลข คืนค่าเป็นข้อความ 'บวก', 'ลบ' หรือ 'ศูนย์'"""
    if num > 0:
        return "บวก"
    elif num < 0:
        return "ลบ"
    else:
        return "ศูนย์"


# --- ข้อที่ 2: ฟังก์ชันตรวจว่าเลขเป็นคู่หรือคี่ ---
def is_even_odd(num):
    """ตรวจสอบว่าตัวเลขเป็นเลขคู่หรือเลขคี่ คืนค่าเป็นข้อความ 'คู่' หรือ 'คี่'"""
    if num % 2 == 0:
        return "คู่"
    else:
        return "คี่"


# --- ข้อที่ 3: ฟังก์ชันรวมผลลัพธ์ เรียกใช้ 2 ฟังก์ชันข้างบนแล้วสร้างประโยคสรุป ---
def describe_number(num):
    """รวมผลตรวจเครื่องหมายและเลขคู่/คี่ แล้วคืนข้อความสรุปของตัวเลขนั้น"""
    sign = is_positive_negative_zero(num)
    parity = is_even_odd(num)
    return f"{num} ผลลัพธ์เป็น{sign} และเป็นเลข {parity}"


# --- ข้อที่ 4: ประโยชน์ที่ได้จากการ refactor คือเรียกซ้ำกี่รอบก็ได้ ไม่ต้อง copy โค้ด ---
print("--- ทดสอบกับตัวเลขหลายค่าในครั้งเดียว ---")
for number in [7, -4, 0, 12, -9]:
    print(describe_number(number))

# --- ข้อที่ 5: ยังรับค่าจากผู้ใช้ได้เหมือนเดิม แต่เหลือโค้ดแค่ 2 บรรทัด ---
print("\n--- ทดสอบกับเลขที่ผู้ใช้กรอกเอง ---")
num = int(input("Enter a number: "))
print(describe_number(num))
````

### สรุปผลการ Refactor

| หัวข้อ | ก่อน Refactor | หลัง Refactor |
| --- | --- | --- |
| การนำกลับมาใช้ซ้ำ | ต้อง copy โค้ดทั้งก้อน | เรียก `describe_number(n)` บรรทัดเดียว |
| การอ่านโค้ด | ตรรกะปนกับ input/print | ชื่อฟังก์ชันบอกหน้าที่ชัดเจน |
| การทดสอบ | ต้องพิมพ์ input ทีละครั้ง | วนลูปทดสอบหลายค่าได้ทันที |
| การแก้ไข | ต้องไล่แก้ทุกจุดที่ copy ไป | แก้ที่ฟังก์ชันเดียว มีผลทุกที่ |

### ตัวอย่างเพิ่มเติม: Refactor Lab 3.2 Guessing Game (Week 3)

เกมทายเลขเดิมมีเงื่อนไขเช็ค สูง/ต่ำ/ถูก อยู่ในลูป และถ้าอยากเล่นด้วยเลขลับตัวใหม่
ต้องแก้ค่าในโค้ดแล้วรันใหม่ทั้งก้อน จึงแยกเป็นฟังก์ชัน `check_guess()` สำหรับตรวจคำตอบ
และ `play_guessing_game()` ที่ใช้ **default parameter** กำหนดจำนวนครั้งที่เดาได้

````python
# ============================================================
# Activity 6.1 - Refactored: Guessing Game (จาก Lab 3.2)
# ============================================================

def check_guess(guess, secret):
    """เปรียบเทียบคำตอบกับเลขลับ คืนข้อความบอกผลว่า low / high / correct"""
    if guess < secret:
        return f"{guess} is too low! Try again."
    elif guess > secret:
        return f"{guess} is too high! Try again."
    else:
        return "Congratulations! You guessed it!"


def play_guessing_game(secret, max_attempts=5):
    """เล่นเกมทายเลข โดยกำหนดจำนวนครั้งที่เดาได้ ถ้าไม่ระบุจะให้เดา 5 ครั้ง (default parameter)"""
    attempts = 0

    while attempts < max_attempts:
        guess = int(input(f"Guess the number - Attempt {attempts + 1}/{max_attempts}: "))
        attempts += 1

        message = check_guess(guess, secret)
        print(message)

        if guess == secret:
            return True          # ทายถูก จบเกมทันที และคืนค่า True

    print(f"Sorry, you ran out of attempts. The secret number was {secret}.")
    return False                 # เดาครบโควตาแล้วยังไม่ถูก คืนค่า False


# เรียกเล่นเกม: ไม่ส่ง max_attempts จะได้ 5 ครั้งตามค่า default
play_guessing_game(42)
````

## ส่วนที่ 2: สำรวจ Python Standard Library

เอกสารอ้างอิง: https://docs.python.org/3/library/ — เลือกมา 3 โมดูลที่น่าสนใจ

````python
# --- โมดูลที่ 1: datetime -> จัดการวันที่และเวลา ---
import datetime

now = datetime.datetime.now()
print(f"เวลาปัจจุบัน: {now}")
print(f"จัดรูปแบบเอง: {now.strftime('%d/%m/%Y %H:%M:%S')}")

# คำนวณวันที่ล่วงหน้า/ย้อนหลังได้ด้วย timedelta
next_week = now + datetime.timedelta(days=7)
print(f"อีก 7 วันคือ: {next_week.strftime('%d/%m/%Y')}")

# หาผลต่างระหว่างวันที่ 2 วัน
exam_day = datetime.date(2026, 12, 1)
days_left = (exam_day - datetime.date.today()).days
print(f"เหลืออีก {days_left} วันจะถึงวันสอบ")

# เอาไปใช้ทำอะไรได้: ระบบเช็คชื่อเข้าเรียนที่บันทึกเวลา, นับถอยหลังวันส่งงาน, คิดอายุจากวันเกิด
````

````python
# --- โมดูลที่ 2: os -> ติดต่อกับระบบปฏิบัติการและไฟล์ ---
import os

print(f"โฟลเดอร์ปัจจุบัน: {os.getcwd()}")
print(f"ไฟล์ในโฟลเดอร์นี้: {os.listdir('.')}")

# เช็คว่าไฟล์มีอยู่จริงไหม ก่อนจะเปิดอ่าน (กันโปรแกรมพัง)
print(f"มีไฟล์ my_utils.py หรือไม่: {os.path.exists('my_utils.py')}")

# ต่อ path แบบไม่ต้องกังวลว่าเป็น Windows หรือ macOS
print(f"ต่อ path: {os.path.join('Week6', 'my_utils.py')}")

# เอาไปใช้ทำอะไรได้: เขียนโปรแกรมจัดระเบียบไฟล์ในเครื่อง, เช็คว่าไฟล์ข้อมูลมีอยู่ก่อนอ่าน, วนอ่านไฟล์ทั้งโฟลเดอร์
````

````python
# --- โมดูลที่ 3: collections -> โครงสร้างข้อมูลเพิ่มเติมนอกเหนือจาก list/dict ปกติ ---
from collections import Counter, defaultdict

# Counter: นับจำนวนของซ้ำให้อัตโนมัติ (ทำงานเหมือน Activity 5.2 แต่เหลือบรรทัดเดียว)
sentence = "python is fun and python is powerful"
word_counts = Counter(sentence.split())
print(f"นับคำ: {word_counts}")
print(f"คำที่เจอบ่อยที่สุด 2 อันดับ: {word_counts.most_common(2)}")

# defaultdict: กำหนดค่าเริ่มต้นให้ key ที่ยังไม่มี ไม่ต้องเขียน if เช็คก่อนทุกครั้ง
scores = defaultdict(list)
scores["Alice"].append(80)      # ไม่ต้องสร้าง scores["Alice"] = [] ก่อน
scores["Alice"].append(95)
scores["Bob"].append(70)
print(f"คะแนนนักเรียน: {dict(scores)}")

# เอาไปใช้ทำอะไรได้: นับความถี่คำในบทความ, นับยอดขายสินค้าแต่ละชิ้น, จัดกลุ่มข้อมูลตามหมวดหมู่
````
