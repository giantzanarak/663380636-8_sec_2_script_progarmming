กิจกรรมที่ 5.2: Word Frequency Counter

```python
# ============================================
# Activity 5.2
# Word Frequency Counter using Dictionary
# ============================================

# 1. รับข้อความจากผู้ใช้
sentence = input("Enter a sentence: ")

# 2. แปลงข้อความเป็นตัวพิมพ์เล็กทั้งหมด
sentence = sentence.lower()

# 3. แยกข้อความออกเป็นคำ
words = sentence.split()

# 4. สร้าง Dictionary เปล่า
word_counts = {}

# 5. วนลูปอ่านคำแต่ละคำ
for word in words:

    # Challenge: ลบเครื่องหมายวรรคตอน
    word = word.strip('.,!?"')

    # 6-8. ตรวจสอบและนับจำนวนคำ
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

# 9. แสดงผลลัพธ์
print("\nWord Frequency:")

for word, count in word_counts.items():
    print(f"{word}: {count}")
```
