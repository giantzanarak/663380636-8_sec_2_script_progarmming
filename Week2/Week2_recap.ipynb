## 1. Control Flow คืออะไร? (30 วินาที)

Control flow คือวิธีที่โปรแกรม **ตัดสินใจ** — แทนที่จะรันจากบนลงล่างทุกครั้งเหมือนเดิม โปรแกรมจะ "แยกเส้นทาง" ตามเงื่อนไข ทุกครั้งที่เขียน `if` เท่ากับสร้างทางแยกให้โค้ดของเรา

---

## 2. ตัวดำเนินการเชิงสัมพันธ์ (Relational Operators) (45 วินาที)

การเปรียบเทียบจะคืนค่าเป็น **บูลีน** (`True`/`False`) เสมอ:

```python
a, b, c = 10, 20, 10

print(a == b)   # False — เท่ากับ
print(a != b)   # True  — ไม่เท่ากับ
print(a < b)    # True  — น้อยกว่า
print(a > b)    # False — มากกว่า
print(a <= c)   # True  — น้อยกว่าหรือเท่ากับ
print(a >= b)   # False — มากกว่าหรือเท่ากับ

# การเปรียบเทียบสตริงจะสนใจตัวพิมพ์เล็ก-ใหญ่ (case-sensitive)
print("python" == "Python")                  # False
print("python".lower() == "Python".lower())  # True
```

---

## 3. ตัวดำเนินการเชิงตรรกะ (Logical Operators) (45 วินาที)

ใช้รวมหรือกลับค่านิพจน์บูลีนด้วย `and`, `or`, `not`:

```python
age = 25
has_license = True

# and -- ต้องเป็น True ทั้งสองฝั่ง
eligible = (age > 21) and has_license
print(eligible)          # True

# or -- เป็น True เพียงฝั่งใดฝั่งหนึ่งก็พอ
can_vote_or_drive = (age >= 18) or has_license
print(can_vote_or_drive) # True

# not -- กลับค่าความจริง
print(not has_license)   # False
```

---

## 4. `if`, `if-else`, `if-elif-else` (1.5 นาที)

```python
# --- if: รันโค้ดเฉพาะเมื่อเงื่อนไขเป็น True ---
temperature = 28
if temperature > 25:
    print("It's a hot day!")

# --- if-else: มีทางเลือกอื่นเมื่อเงื่อนไขเป็น False ---
number = 7
if number % 2 == 0:
    print(f"{number} is even.")
else:
    print(f"{number} is odd.")

# --- if-elif-else: เงื่อนไขที่แยกจากกันชัดเจน ---
# โปรแกรมจะหยุดตรวจที่เงื่อนไข True ตัวแรกที่เจอ ดังนั้นลำดับสำคัญมาก!
score = 85
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"
print(f"Grade: {grade}")   # Grade: B
```

---

## 5. เงื่อนไขซ้อนกัน (Nested Conditionals) (45 วินาที)

บล็อก `if` สามารถมี `if` ซ้อนอยู่ข้างในได้ — มีประโยชน์ แต่ถ้าซ้อนลึกเกินไป ความอ่านง่ายจะลดลงอย่างรวดเร็ว

```python
age = 17
has_license = True

if age >= 18:
    if has_license:
        print("You can drive legally.")
    else:
        print("You need a license to drive.")
else:
    print("You are too young to drive.")
# ผลลัพธ์: "You are too young to drive."
```

## 6. แอบดูล่วงหน้า: Comprehensions (30 วินาที)

บทเรียนถัดไปจะย่อ loop + เงื่อนไขให้เหลือบรรทัดเดียว — ตัวอย่างคร่าว ๆ:

```python
# แบบ loop ปกติ
evens = []
for x in range(1, 11):
    if x % 2 == 0:
        evens.append(x)

# ผลลัพธ์เดียวกัน แต่เขียนแบบ list comprehension
evens = [x for x in range(1, 11) if x % 2 == 0]
print(evens)   # [2, 4, 6, 8, 10]
```

---"""
learning_log_week2.py
----------------------
Week 2 learning-log tool: same journaling habit as Week 1,
now built using THIS week's concepts -- relational operators,
logical operators, and if / elif / else -- so the tool itself
is a working example of control flow.

Run it with:
    python learning_log_week2.py
"""

import logging
from datetime import datetime

logging.basicConfig(
    filename="learning_log.txt",
    level=logging.INFO,
    format="%(asctime)s | %(message)s",
    datefmt="%Y-%m-%d %H:%M",
)


def get_entry():
    """Ask what was studied today and how confident the student feels."""
    topic = input("Topic studied today (e.g. 'if-elif-else'): ")
    confidence_str = input("Confidence level, 1-5: ")
    confidence = int(confidence_str)   # input() always returns a string
    struggled = input("Did you get stuck on anything? (yes/no): ").lower()
    return topic, confidence, struggled


def classify_confidence(confidence, struggled):
    """
    Turn a raw score into a plain-language status using
    if-elif-else -- mutually exclusive, order matters.
    """
    # relational operators (>=, <) + logical operator (and)
    if confidence >= 4 and struggled == "no":
        status = "Solid -- ready for the next topic."
    elif confidence >= 3:
        status = "Getting there -- one more practice round."
    elif confidence >= 1:
        status = "Needs review -- re-read the recap and redo the snippets."
    else:
        status = "Invalid confidence score."
    return status


def log_entry(topic, confidence, struggled, status):
    entry = f"Topic: {topic} | Confidence: {confidence}/5 | Stuck: {struggled} | Status: {status}"
    logging.info(entry)
    print("\nSaved to learning_log.txt:")
    print(entry)


def main():
    print("=== Week 2 Learning Log: Control Flow ===")
    print(f"Today's date: {datetime.now():%Y-%m-%d}\n")

    topic, confidence, struggled = get_entry()
    status = classify_confidence(confidence, struggled)
    log_entry(topic, confidence, struggled, status)

    # not -- inverts a boolean, another Week 2 concept
    needs_review = not (confidence >= 4)
    if needs_review:
        print("\nTip: schedule 10 minutes tomorrow to re-trace the Snippet A/B/C examples by hand.")


if __name__ == "__main__":
    main()
