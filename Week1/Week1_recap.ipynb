## 4. ตัวแปรและชนิดข้อมูล (1.5 นาที)

Python เป็นภาษาที่กำหนดชนิดข้อมูลแบบไดนามิก (dynamically typed) — ไม่ต้องประกาศชนิดข้อมูลเอง Python จะอนุมานให้เอง

```python
# --- จำนวนเต็ม (Integers) ---
age = 30
print("My age is:", age)
print("Type of age:", type(age))     # <class 'int'>

# --- ทศนิยม (Floats) ---
price = 19.99
print("Type of price:", type(price)) # <class 'float'>

# --- สตริง (Strings) ---
name = "Alice"
greeting = 'Hello'                   # ใช้เครื่องหมายคำพูดเดี่ยวหรือคู่ก็ได้
print(greeting + ", " + name + "!")  # การต่อสตริงด้วย +

# --- บูลีน (Booleans) ---
is_student = True
print("Type of is_student:", type(is_student))  # <class 'bool'>
```

**กฎการตั้งชื่อ (PEP 8):** ใช้ `snake_case`, เริ่มต้นด้วยตัวอักษรหรือ `_` และห้ามใช้คำสงวน (keyword) ของ Python เป็นชื่อตัวแปร

## 5. การรับ/แสดงผลข้อมูล (1 นาที)

`input()` จะ **คืนค่าเป็นสตริงเสมอ** — ต้องแปลงชนิดข้อมูลก่อนนำไปคำนวณ

```python
user_name = input("Enter your name: ")
print("Hello,", user_name + "!")

age_str = input("Enter your age: ")
user_age = int(age_str)              # แปลง str -> int
print("In 5 years you'll be", user_age + 5)

# f-strings — วิธีจัดรูปแบบข้อความที่ทันสมัยและอ่านง่าย
first_name = "Ada"
last_name = "Lovelace"
full_name = first_name + " " + last_name
print(f"Hello, {full_name}!")
```

---"""
learning_log_th.py
--------------------
โปรเจกต์เล็ก ๆ สำหรับ "สัปดาห์ที่ 1": บันทึกสิ่งที่เรียนรู้วันนี้

ใช้เฉพาะแนวคิดของสัปดาห์ที่ 1 — print(), คอมเมนต์, ตัวแปร,
input(), การต่อสตริง/f-strings, และการเขียนไฟล์เบื้องต้น —
พร้อมโมดูล `logging` ของ Python เพื่อเริ่มสร้างนิสัย
"จดสิ่งที่เรียนรู้" ตั้งแต่วันแรก

รันด้วยคำสั่ง:
    python learning_log_th.py
"""

import logging
from datetime import datetime

# --- ตั้งค่า logging: บันทึกลงไฟล์ learning_log.txt พร้อมเวลาที่บันทึก ---
logging.basicConfig(
    filename="learning_log.txt",
    level=logging.INFO,
    format="%(asctime)s | %(message)s",
    datefmt="%Y-%m-%d %H:%M",
)

def get_entry():
    """ถามผู้ใช้ว่าวันนี้เรียนอะไรมา คืนค่าเป็นสตริง"""
    topic = input("หัวข้อที่เรียนวันนี้ (เช่น 'print() และตัวแปร'): ")
    confidence = input("มั่นใจแค่ไหน ให้คะแนน 1-5: ")

    # input() คืนค่าเป็นสตริงเสมอ -- ต้องแปลงชนิดข้อมูลก่อนคำนวณ
    confidence_score = int(confidence)

    note = input("มีจุดไหนที่ยังงงอยู่ไหม (หรือพิมพ์ 'none'): ")
    return topic, confidence_score, note


def log_entry(topic, confidence_score, note):
    """บันทึกข้อมูลหนึ่งรายการโดยใช้ f-string"""
    entry = f"หัวข้อ: {topic} | ความมั่นใจ: {confidence_score}/5 | จุดที่งง: {note}"
    logging.info(entry)
    print("\nบันทึกลงไฟล์ learning_log.txt แล้ว:")
    print(entry)


def main():
    print("=== บันทึกการเรียนรู้ สัปดาห์ที่ 1 ===")
    print(f"วันที่วันนี้: {datetime.now():%Y-%m-%d}\n")

    topic, confidence_score, note = get_entry()
    log_entry(topic, confidence_score, note)

    # ข้อคิดเล็ก ๆ เกี่ยวกับการติดตามตัวเองอย่างมีความรับผิดชอบ:
    # ความมั่นใจต่ำไม่ใช่ปัญหา แต่คือข้อมูลที่บอกว่าควรทบทวนเรื่องไหนเพิ่ม
    if confidence_score <= 2:
        print("\nคำแนะนำ: ความมั่นใจ 1-2 แปลว่าหัวข้อนี้ควรทบทวนอีกรอบ")
        print("ลองอ่านสรุปอีกครั้ง แล้วลองรันโค้ดตัวอย่างใหม่ดู")


if __name__ == "__main__":
    main()
